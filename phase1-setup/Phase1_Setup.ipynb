## Cell 1 - Mount Drive and create project folders

```python
from google.colab import drive
import os

drive.mount('/content/drive')

# project base directory
base       = '/content/drive/MyDrive/FakeNewsThesis'
data_dir   = base + '/data'
model_dir  = base + '/model'
charts_dir = base + '/charts'

# create folders if they don't exist
for folder in [base, data_dir, model_dir, charts_dir]:
    os.makedirs(folder, exist_ok=True)
    print('ready:', folder)

print()
print('project structure created')
print('base     :', base)
print('data     :', data_dir)
print('model    :', model_dir)
print('charts   :', charts_dir)
```

## Cell 2 - Check GPU and Python environment

```python
import torch
import sys

print('Python version :', sys.version[:20])
print('PyTorch version:', torch.__version__)
print('CUDA available :', torch.cuda.is_available())

if torch.cuda.is_available():
    print('GPU            :', torch.cuda.get_device_name(0))
    print('GPU memory     :', round(torch.cuda.get_device_properties(0).total_memory / 1e9, 1), 'GB')
else:
    print('WARNING: no GPU found - go to Runtime -> Change runtime type -> T4 GPU')
```

## Cell 3 - Install required packages

```python
import subprocess

print('installing packages...')

# install tokenizers first to avoid build-from-source error
subprocess.run(['pip', 'install', 'tokenizers==0.19.1', '-q'], check=False)
subprocess.run(['pip', 'install', 'transformers==4.41.0', '-q'], check=False)
subprocess.run(['pip', 'install', 'kafka-python', 'plotly', 'dash', 'pyngrok', '-q'], check=False)

print('done')

# verify installs
import transformers
print('transformers version:', transformers.__version__)
```

## Cell 4 - Test DistilBERT tokenizer loads correctly

```python
from transformers import AutoTokenizer

print('loading DistilBERT tokenizer...')
tokenizer = AutoTokenizer.from_pretrained('distilbert-base-uncased')

# quick test
test = 'Scientists confirm the new vaccine is safe and effective.'
tokens = tokenizer(test, return_tensors='pt', max_length=128, truncation=True)
print('test sentence  :', test)
print('token count    :', tokens['input_ids'].shape[1])
print()
print('environment setup complete - ready for Phase 2')
```
